# Basic String Operations:

## Concatenation:


````python
a = "Hello"
b = "world"

print("Concate: ", a + ' ' + b)
````

## Length of string:


````python
s = "python"

print("Length of the string: ", len(s))
````

## Accessing characters (Indexing):


````python
name = "Programming is fun!"

print(name[0])
print(name[1])
print(name[5])
````

## Slicing:


````python
name = "raselsarker"

print(name[1:])
print(name[1:4])
print(name[3:-1])
print(name[:-1])
````

## String Methods (Useful Built-in Functions):


## Uppercase / Lowercase:


````python
a = "PYTHON"

print(a.lower())
````

````python
b = "machine learning"

print(b.upper())
````

## Replace:


````python
name = "Machine Learning".replace("Machine", "Deep")

print(name)
````

## Split and Join:


````python
"a, b, c".split(",")
"#".join(["a", "b", "c"])
````

## Strip (Remove whitespace):


````python
"hello  ".strip()
````

## Startswith / Endswith:


````python
"machine learning".startswith("ma")
````

````python
"machine learning".endswith("nb")
````

## Find and Count:


````python
"rootsofintelligence".count("i")
````

````python
"rootsofintelligence".find("o")
````

## Formatting Strings:  
- f-strings:


````python
name = "Rasel sarker"
age = 25

print(f"your name is {name}. and you are {age} years old.")
````

## String Check Methods (Return True/False):

````python
"root".isalpha()
"root".isdigit()
````

````python
"1256".isdigit()
````

````python
"python145".isalnum()
````

````python
"Strings".istitle()
````
