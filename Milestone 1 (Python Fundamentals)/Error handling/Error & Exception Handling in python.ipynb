5. **Type Error:** Wrong type of data used.

    <h2 style="text-align: center; color: #333366;">Runtime Errors (Built-in Exceptions)</h2> 


---

1. **ZeroDivisionError:** → 0 দিয়ে ভাগ korar try kora. 
2. **ValueError:** → ভুল value input দেওয়া (e,g: int("abc"))
3. **IndexError:** → list বা tuple এর বাইরে থেকে উপাদান অ্যাক্সেস করতে চেষ্টা korlay.
4. **KeyError:** → dictionary তে key না পাওয়া গেলে
5. **FileNotFoundError:** → trying to access a file from somewhere but the file doesn't exist.
6. **TypeError:** → When we use the wrong data type(যেমন "abc" + 10)
7. **AttributeError:** → try to access an attribute of an object that does not exist.
8. **ImportError/ModuleNotFoundError:** → module import problem
9. **NameError:** → variable define করা নাই
10. **MemoryError:** → Out of memory

    <h2 style="text-align: center; color: #333366;">Error Handling methods</h2> 


---


1. **try:** যেখানে আপনি কোডের এমন অংশ লিখেন যা error ঘটাতে পারে। 
2. **except:** error ঘটলে সেটি ধরে এবং handle করতে।
3. **else:**  যখন কোন error না ঘটে, তখন এটি execute হয়।
4. **finally:** এটি always execute হয়, error আসলেও হয় না আসলেও হয়।
5. **raise:** নিজের custom exceptions তৈরি করার জন্য ব্যবহার করা হয়।

    <h2 style="text-align: center; color: #333366;">File handling in Python</h2> 


---


1. **r** → Read (default)
2. **w** → Write (will erase previous data and write new)
3. **a** → Append (will add new to old)
4. **b** → Binary mode (for image, video etc. files)
5. **x** → Create (will give an error if the file already exists)
