````python
a = {1, 2, 3, 4}
b = {2, 5, 9, 10, 23}

# 1, 3, 4, 5, 9, 23
# 1 5 3 

print("Intersection is", a.intersection(b))
print("difference is", a.difference(b))
print("symmetric_difference is", a.symmetric_difference(b))
print("issubset is", a.issubset(b))
print("proper subset: ", a != b)
print("super subset: ", a.issuperset(b))
print(len(a))
print(len(b))
````

````python
a = {10, 2, 3, 4}
b = {2, 5, 9, 10, 23}

print(a.intersection(b))
print(sorted(a & b))
print(sorted(a.intersection(b)))
````
