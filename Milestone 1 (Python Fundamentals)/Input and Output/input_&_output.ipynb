```python
num = input("Enter your number separated by space: ").split()

num_List = [int(n) for n in num]
print("Your List is: ", num_List)
```

```python
num = input("Enter your number separated by space: ").split()

num_List = tuple(int(n) for n in num)
print("Your tuple is: ", num_List)
```

```python
num = input("Enter your number separated by space: ")

num_List = set(int(n) for n in num.split())
print("Your set is: ", num_List)
```

```python
num = input("Enter your number to this format key:value and key:value pairs: ")

num_List = dict(n.split(':') for n in num.split())
print("Your dictionary is: ", num_List)
```
